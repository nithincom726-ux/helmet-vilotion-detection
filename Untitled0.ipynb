```python
from ultralytics import YOLO
import cv2

model = YOLO("/content/runs/detect/train/weights/best.pt")

image_path = "/content/nohelmet.jpg"

results = model(image_path, conf=0.5)

img = cv2.imread(image_path)

for result in results:
    boxes = result.boxes

    for box in boxes:
        cls = int(box.cls[0])
        conf = float(box.conf[0])

        x1, y1, x2, y2 = map(int, box.xyxy[0])

        class_name = model.names[cls]

        if class_name == "no_helmet":
            label = "HELMET VIOLATION"
            color = (0, 0, 255)  # Red
        elif class_name == "helmet":
            label = "NO VIOLATION"
            color = (0, 255, 0)  # Green
        else:
            continue

        cv2.rectangle(img, (x1,y1), (x2,y2), color, 2)

        cv2.putText(
            img,
            label,
            (x1, y1-10),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.7,
            color,
            2
        )

cv2.imwrite("/content/violation_result.jpg", img)

print("Saved!")
```

```python
from ultralytics import YOLO
import cv2

# Load trained model
model = YOLO("/content/runs/detect/train/weights/best.pt")

# Your video path
video_path = "/content/5638-183850414_medium.mp4"

cap = cv2.VideoCapture(video_path)

width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
fps = int(cap.get(cv2.CAP_PROP_FPS))

output_path = "/content/helmet_violation_output.mp4"

out = cv2.VideoWriter(
    output_path,
    cv2.VideoWriter_fourcc(*"mp4v"),
    fps,
    (width, height)
)

while True:
    ret, frame = cap.read()

    if not ret:
        break

    results = model(frame, conf=0.35)

    for result in results:
        for box in result.boxes:

            cls = int(box.cls[0])
            confidence = float(box.conf[0])

            x1, y1, x2, y2 = map(int, box.xyxy[0])

            class_name = model.names[cls]

            if class_name == "no_helmet":
                label = "HELMET VIOLATION"
                color = (0, 0, 255)

            elif class_name == "helmet":
                label = "NO VIOLATION"
                color = (0, 255, 0)

            else:
                continue

            cv2.rectangle(
                frame,
                (x1, y1),
                (x2, y2),
                color,
                2
            )

            cv2.putText(
                frame,
                label,
                (x1, y1 - 10),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.7,
                color,
                2
            )

    out.write(frame)

cap.release()
out.release()

print("✅ Video processing completed")
print("Saved at:", output_path)
```

```python
from ultralytics import YOLO
import cv2
import os

model = YOLO("/content/runs/detect/train/weights/best.pt")

video_path = "/content/5638-183850414_medium.mp4"

cap = cv2.VideoCapture(video_path)

os.makedirs("/content/violations", exist_ok=True)

count = 0

while True:

    ret, frame = cap.read()

    if not ret:
        break

    results = model(frame, conf=0.35)

    violation_found = False

    for result in results:

        for box in result.boxes:

            cls = int(box.cls[0])
            name = model.names[cls]

            if name == "no_helmet":

                violation_found = True

                x1,y1,x2,y2 = map(int, box.xyxy[0])

                cv2.rectangle(
                    frame,
                    (x1,y1),
                    (x2,y2),
                    (0,0,255),
                    3
                )

                cv2.putText(
                    frame,
                    "HELMET VIOLATION",
                    (x1,y1-10),
                    cv2.FONT_HERSHEY_SIMPLEX,
                    0.8,
                    (0,0,255),
                    2
                )


    if violation_found:
        count += 1
        cv2.imwrite(
            f"/content/violations/violation_{count}.jpg",
            frame
        )


cap.release()

print("Total violations saved:", count)
```

```python
%%writefile app.py

import streamlit as st
from ultralytics import YOLO
from PIL import Image
import cv2
import numpy as np

# Load model
model = YOLO("/content/runs/detect/train/weights/best.pt")

st.title("🚦 Helmet Violation Detection System")

st.write("Upload an image to detect helmet violations")

uploaded_file = st.file_uploader(
    "Choose an image",
    type=["jpg","jpeg","png"]
)

if uploaded_file:

    image = Image.open(uploaded_file)

    st.image(
        image,
        caption="Uploaded Image",
        use_container_width=True
    )

    img = np.array(image)

    results = model(img, conf=0.35)

    for result in results:

        for box in result.boxes:

            cls = int(box.cls[0])
            name = model.names[cls]

            x1,y1,x2,y2 = map(int, box.xyxy[0])

            if name == "no_helmet":

                color = (255,0,0)
                label = "HELMET VIOLATION 🚨"

            elif name == "helmet":

                color = (0,255,0)
                label = "NO VIOLATION ✅"

            else:
                continue


            cv2.rectangle(
                img,
                (x1,y1),
                (x2,y2),
                color,
                3
            )

            cv2.putText(
                img,
                label,
                (x1,y1-10),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.7,
                color,
                2
            )


    st.subheader("Detection Result")

    st.image(
        img,
        channels="RGB",
        use_container_width=True
    )
```
